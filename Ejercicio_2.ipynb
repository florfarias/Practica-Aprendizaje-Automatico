Importamos y cargamos los datos con la estructura adecuada para manipularlos.

```python
import pandas as pd
import matplotlib.pyplot as plt

automoviles = pd.read_csv("/content/Automobile.csv")

automoviles.head()
```

Creamos un gráfico de dispersión para visualizar las relaciones entre los pares de características numéricas `mpg` y `weight`.

```python
plt.scatter(automoviles["weight"], automoviles["mpg"])

plt.xlabel("Peso del automóvil")
plt.ylabel("Millas por galón")

plt.show()
```

Generamos un histograma del consumo de combustible (`mpg`) para analizar su distribución.

```python
plt.hist(automoviles["mpg"], bins=10, edgecolor="yellow")

plt.xlabel("Millas por galón (mpg)")
plt.ylabel("Cantidad de automóviles")

plt.show()
```

 Graficamos un diagrama de caja (boxplot) de las características `cylinders` y `origin` para identificar posibles valores atípicos.

*Para convertir las columnas a datos númericos le tuve que pedir ayuda a la IA porque no sabia cómo hacer que aparezca la caja de origen ya que el dato era el continente y no un numero, pero no se si está bien resuelto para interpretación.*

```python
# Convertimos las columnas a datos numéricos
automoviles = pd.read_csv("/content/Automobile.csv")
automoviles["cylinders"] = pd.to_numeric(automoviles["cylinders"],errors="coerce")
origen_numerico = automoviles["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})
```

```python
plt.boxplot([automoviles["cylinders"].dropna(),origen_numerico.dropna()])

plt.xticks([1, 2], ["Cilindros", "Origen"])
plt.ylabel("Valores")

plt.show()
```

Realizamos un gráfico de barras que muestre la distribución de automóviles según su año de modelo (`model_year`) y origen (`origin`).

```python
distribucion = automoviles.groupby(["model_year", "origin"]).size().unstack()

distribucion
```

```python
distribucion.plot(kind="bar")

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")

plt.show()
```
